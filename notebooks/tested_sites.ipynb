# Total tested count - including NaNs

```python
import pandas as pd
from pathlib import Path
from glob import glob

directory = '/scratch/gpfs/AMOELLER/diet_manip/AlleleFlux_mapq20/longitudinal/significance_tests/preprocessed_between_groups_pre_end-fat_control'
files = sorted(glob(f'{directory}/*_preprocessed.tsv.gz'))

print(f'Found {len(files)} preprocessed files')
print()

total_positions = set()
per_mag_counts = {}

for f in files:
    mag_id = Path(f).name.replace('_allele_frequency_changes_mean_preprocessed.tsv.gz', '')
    df = pd.read_csv(f, sep='\t', usecols=['contig', 'position'])
    unique_pos = df.drop_duplicates(subset=['contig', 'position'])
    n = len(unique_pos)
    per_mag_counts[mag_id] = n
    for _, row in unique_pos.iterrows():
        total_positions.add((row['contig'], row['position']))

print(f'Total unique (contig, position) across all MAGs: {len(total_positions):,}')
print(f'Sum of per-MAG unique positions: {sum(per_mag_counts.values()):,}')
print()
print('Per-MAG unique position counts (top 10):')
for mag, count in sorted(per_mag_counts.items(), key=lambda x: -x[1])[:10]:
    print(f'  {mag}: {count:,}')
print(f'  ... ({len(per_mag_counts)} MAGs total)')

```

# Tested sites per MAG

```python
import re
import pandas as pd
from pathlib import Path
from glob import glob

# =========================
# Inputs
# =========================
single_sample_path = "/scratch/gpfs/AMOELLER/diet_manip/AlleleFlux_mapq20/longitudinal/p_value_summary/pre_end-fat_control/p_value_summary_single_sample_pre_end.tsv"
paired_path        = "/scratch/gpfs/AMOELLER/diet_manip/AlleleFlux_mapq20/longitudinal/p_value_summary/pre_end-fat_control/p_value_summary_two_sample_paired_pre_end.tsv"

preprocessed_between_dir = "/scratch/gpfs/AMOELLER/diet_manip/AlleleFlux_mapq20/longitudinal/significance_tests/preprocessed_between_groups_pre_end-fat_control"
preprocessed_within_dir  = "/scratch/gpfs/AMOELLER/diet_manip/AlleleFlux_mapq20/longitudinal/significance_tests/preprocessed_within_groups_pre_end-fat_control"

out_path = "/scratch/gpfs/AMOELLER/sidd/diet_manip/tested_sites/sites_tested_per_SGB_within_and_between.tsv"

# =========================
# Helpers
# =========================
def count_unique_sites_per_mag(df: pd.DataFrame, mag_col="mag_id") -> pd.Series:
    """Counts unique (contig, position) per mag_id."""
    return (
        df[[mag_col, "contig", "position"]]
        .drop_duplicates()
        .groupby(mag_col)
        .size()
    )

def mag_id_from_between_preprocessed_filename(fname: str) -> str:
    """
    Example:
      SLG443_DASTool_bins_SLG443_bin.96_allele_frequency_changes_mean_preprocessed.tsv.gz
    -> SLG443_DASTool_bins_SLG443_bin.96
    """
    return re.sub(r"_allele_frequency_changes_mean_preprocessed\.tsv\.gz$", "", fname)

def count_preprocessed_between(preprocessed_dir: str, label: str) -> pd.Series:
    files = sorted(glob(f"{preprocessed_dir}/*_preprocessed.tsv.gz"))
    if len(files) == 0:
        raise FileNotFoundError(f"No *_preprocessed.tsv.gz files found in: {preprocessed_dir}")

    rows = []
    for f in files:
        name = Path(f).name
        mag_id = mag_id_from_between_preprocessed_filename(name)
        tmp = pd.read_csv(f, sep="\t", usecols=["contig", "position"])
        n_sites = tmp.drop_duplicates(["contig", "position"]).shape[0]
        rows.append((mag_id, n_sites))

    return pd.DataFrame(rows, columns=["mag_id", label]).set_index("mag_id")[label]

def parse_within_filename(fname: str):
    """
    Examples:
      SLG887_DASTool_bins_SLG887_bin.36_control_allele_frequency_changes_mean_zeros_processed.tsv.gz
      SLG887_DASTool_bins_SLG887_bin.36_fat_allele_frequency_changes_mean_zeros_processed.tsv.gz
    Returns (mag_id, group) where group in {"fat","control"}.
    """
    m = re.match(
        r"^(?P<mag>.+)_(?P<grp>fat|control)_allele_frequency_changes_mean_zeros_processed\.tsv\.gz$",
        fname
    )
    if not m:
        raise ValueError(f"Unexpected within-preprocessed filename format: {fname}")
    return m.group("mag"), m.group("grp")

def count_preprocessed_within(preprocessed_dir: str):
    """
    Returns two Series indexed by mag_id:
      - n_preprocessed_within_fat
      - n_preprocessed_within_control
    """
    files = sorted(glob(f"{preprocessed_dir}/*_zeros_processed.tsv.gz"))
    if len(files) == 0:
        raise FileNotFoundError(f"No *_zeros_processed.tsv.gz files found in: {preprocessed_dir}")

    rows = []
    for f in files:
        name = Path(f).name
        mag_id, grp = parse_within_filename(name)

        tmp = pd.read_csv(f, sep="\t", usecols=["contig", "position"])
        n_sites = tmp.drop_duplicates(["contig", "position"]).shape[0]
        rows.append((mag_id, grp, n_sites))

    df_counts = pd.DataFrame(rows, columns=["mag_id", "group", "n_sites"])

    within_fat = (
        df_counts[df_counts["group"] == "fat"]
        .set_index("mag_id")["n_sites"]
        .rename("n_preprocessed_within_fat")
    )
    within_control = (
        df_counts[df_counts["group"] == "control"]
        .set_index("mag_id")["n_sites"]
        .rename("n_preprocessed_within_control")
    )
    return within_fat, within_control

# =========================
# 1) single_sample_tTest per SGB (fat, control, UNION across fat+control)
# =========================
df_single_all = pd.read_csv(single_sample_path, sep="\t")
df_single = df_single_all[df_single_all["test_type"] == "single_sample_tTest"].copy()

fat_sites = count_unique_sites_per_mag(
    df_single[df_single["group_analyzed"] == "fat"]
).rename("n_single_sample_tTest_fat")

control_sites = count_unique_sites_per_mag(
    df_single[df_single["group_analyzed"] == "control"]
).rename("n_single_sample_tTest_control")

# UNION: unique (contig, position) per SGB across fat+control
union_sites = count_unique_sites_per_mag(df_single).rename("n_single_sample_tTest_union")

# =========================
# 2) two_sample_paired_tTest per SGB
# =========================
df_paired_all = pd.read_csv(paired_path, sep="\t")
paired_sites = count_unique_sites_per_mag(
    df_paired_all[df_paired_all["test_type"] == "two_sample_paired_tTest"]
).rename("n_two_sample_paired_tTest")

# =========================
# 3) preprocessed: between + within (fat/control)
# =========================
pre_between = count_preprocessed_between(preprocessed_between_dir, "n_preprocessed_between_groups")
pre_within_fat, pre_within_control = count_preprocessed_within(preprocessed_within_dir)

# =========================
# 4) Combine into one per-SGB table (+ TOTAL row)
# =========================
sites_per_sgb = (
    pd.concat(
        [fat_sites, control_sites, union_sites, pre_within_fat, pre_within_control, paired_sites, pre_between],
        axis=1
    )
    .fillna(0)
    .astype(int)
    .rename_axis("SGB")
    .reset_index()
)

# Add TOTAL row (column sums across SGBs)
numeric_cols = [c for c in sites_per_sgb.columns if c != "SGB"]
totals = sites_per_sgb[numeric_cols].sum()

total_row = pd.DataFrame([{"SGB": "TOTAL", **totals.to_dict()}])
sites_per_sgb_total = pd.concat([sites_per_sgb, total_row], ignore_index=True)

# =========================
# 5) Print totals + write
# =========================
print("TOTAL single_sample_tTest fat             :", int(totals["n_single_sample_tTest_fat"]))
print("TOTAL single_sample_tTest control         :", int(totals["n_single_sample_tTest_control"]))
print("TOTAL single_sample_tTest union           :", int(totals["n_single_sample_tTest_union"]))  # should match your 517,704
print("TOTAL two_sample_paired_tTest             :", int(totals["n_two_sample_paired_tTest"]))
print("TOTAL preprocessed BETWEEN groups         :", int(totals["n_preprocessed_between_groups"]))
print("TOTAL preprocessed WITHIN fat             :", int(totals["n_preprocessed_within_fat"]))
print("TOTAL preprocessed WITHIN control         :", int(totals["n_preprocessed_within_control"]))

sites_per_sgb_total.to_csv(out_path, sep="\t", index=False)
print("Wrote:", out_path)

sites_per_sgb_total.tail()
```
