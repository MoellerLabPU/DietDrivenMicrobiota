```python
import pandas as pd
```

```python
# ----------------------------
# Paths
# ----------------------------
pval_path = "/scratch/gpfs/AMOELLER/diet_manip/AlleleFlux_mapq20/longitudinal/p_value_summary/pre_end-fat_control/p_value_summary_two_sample_paired_pre_end.tsv"
cog_path  = "/scratch/gpfs/AMOELLER/sidd/diet_manip/reCOGnizer_results.tsv"
```

```python
# ----------------------------
# Load p-values summary
# ----------------------------
df_summary = pd.read_csv(pval_path, sep="\t")

# Optional: restrict to a single test type
df_summary = df_summary.loc[df_summary["test_type"].eq("two_sample_paired_tTest")].copy()
df_summary
```

```python
# ----------------------------
# Explode gene_id so it's ONE gene per row
# Handles cells like: "geneA ,geneB" or "geneA, geneB"
# ----------------------------
df_exploded = df_summary.copy()

# Make a list column from gene_id (robust to NaN)
gene_series = df_exploded["gene_id"].astype("string")

# Split on commas with optional whitespace around them
df_exploded["gene_id_single"] = gene_series.str.split(r"\s*,\s*")

# Explode into one row per gene
df_exploded = df_exploded.explode("gene_id_single", ignore_index=True)

# Clean whitespace; drop empties/NAs
df_exploded["gene_id_single"] = df_exploded["gene_id_single"].astype("string").str.strip()
df_exploded = df_exploded.loc[
    df_exploded["gene_id_single"].notna() & (df_exploded["gene_id_single"] != "")
].copy()

df_exploded
```

```python
# ----------------------------
# Load + prep COG/reCOGnizer table (only needed cols)
# ----------------------------
usecols = ["qseqid", "Protein description", "DB description", "DB ID", "CDD ID", "General functional category", "Functional category"]
df_cog = pd.read_csv(cog_path, sep="\t", usecols=usecols)

# Extract gene_id from qseqid and normalize whitespace
df_cog["gene_id_single"] = (
    df_cog["qseqid"].astype("string").str.split("_#", n=1).str[0].str.strip()
)
df_cog
```

```python
annot_cols = ["Protein description", "DB description", "DB ID", "CDD ID", "General functional category", "Functional category"]
placeholder = "-"  # or "" if you truly want blank, but "-" is clearer

# Fill missing values so comma-separated lists stay aligned
for c in annot_cols:
    df_cog[c] = df_cog[c].astype("string").fillna(placeholder)
    df_cog[c] = df_cog[c].replace("", placeholder)

df_cog
```

```python
# Printing lines that are the exact duplicates
df_check = df_cog  # or df_cog_subset

dup_mask = df_check.duplicated(subset=["gene_id_single"] + annot_cols, keep=False)

dups = (
    df_check.loc[dup_mask, ["gene_id_single"] + annot_cols]
            .sort_values(["gene_id_single"] + annot_cols)
)

print(len(dups), "duplicate rows (including all copies)")
dups.head(n=20)
```

```python
# Filter COG table with only the required genes
need = set(df_exploded["gene_id_single"].dropna().unique())
df_cog = df_cog[df_cog["gene_id_single"].isin(need)].copy()
df_cog
```

```python
# (Optional but recommended) drop exact duplicate annotation rows per gene
df_cog = df_cog.drop_duplicates(subset=["gene_id_single"] + annot_cols)

# Aggregate to one row per gene (comma-separated lists)
cog_agg = (
    df_cog.groupby("gene_id_single", sort=False, as_index=False)[annot_cols]
          .agg(lambda s: ", ".join(s.tolist()))
)
cog_agg
```

```python
# ----------------------------
# Merge annotations onto exploded p-value table
# ----------------------------
df_exploded_annot = df_exploded.merge(cog_agg, on="gene_id_single", how="left")

# If you prefer, drop the original gene_id column (which may contain comma-separated genes)
# df_exploded_annot = df_exploded_annot.drop(columns=["gene_id"])
df_exploded_annot.to_csv("/scratch/gpfs/AMOELLER/sidd/diet_manip/p_value_summary_reCOGnizer.tsv", sep="\t", index=False)
df_exploded_annot
```

```python
# Make missing values explicit so joins don't "skip" them
for c in annot_cols:
    cog[c] = cog[c].fillna(placeholder)
    cog[c] = cog[c].replace("", placeholder)  # treat empty strings as missing too

# OPTIONAL: drop exact duplicate hit-rows (keeps alignment because it's row-level)
cog = cog.drop_duplicates(subset=["gene_id"] + annot_cols)
cog
```

```python
annot_cols = ["Protein description", "DB description", "DB ID", "CDD ID"]
placeholder = "-"

# start from a clean copy
cog_pre = df_cog_subset.copy()

# fill missing (this is the state you actually deduplicate on)
for c in annot_cols:
    cog_pre.loc[:, c] = cog_pre[c].fillna(placeholder).replace("", placeholder)

# Rows that drop_duplicates(keep="first") will remove:
to_drop = cog_pre.duplicated(subset=["gene_id"] + annot_cols, keep="first")
print("rows before:", len(cog_pre))
print("rows that would be dropped:", int(to_drop.sum()))
print("rows after:", len(cog_pre) - int(to_drop.sum()))

# Show the duplicate groups (all copies)
dups_all = (
    cog_pre.loc[cog_pre.duplicated(subset=["gene_id"] + annot_cols, keep=False),
                ["gene_id"] + annot_cols]
          .sort_values(["gene_id"] + annot_cols)
)
dups_all.head(50)
```

```python

# Aggregate: one row per gene_id, comma-separated lists per column
cog_agg = (
    cog.groupby("gene_id", sort=False, as_index=False)[annot_cols]
       .agg(lambda s: ", ".join(s.tolist()))
)

# Merge onto your summary table (left join keeps all summary rows)
df_merged = df_summary_subset.merge(cog_agg, on="gene_id", how="left")
```
